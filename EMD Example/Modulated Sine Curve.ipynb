```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import hilbert

# ------------------------------------------------------------
# Modulated sine curve
# x(t) = A(t) * sin(omega * t)
# ------------------------------------------------------------
n = 2000
t = np.linspace(0, 1, n, endpoint=False)

# Slowly varying amplitude
A = 1.0 + 0.4 * np.cos(2 * np.pi * 2 * t)

# Carrier
omega = 2 * np.pi * 18

# Signal
x = A * np.sin(omega * t)

# Hilbert transform via analytic signal
z = hilbert(x)
Hx = np.imag(z)

# ------------------------------------------------------------
# Transparent plot, no insets
# ------------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 4), facecolor="none")
ax.set_facecolor("none")

ax.plot(t, x, lw=2)
ax.plot(t, Hx, lw=2)

ax.set_xlabel("t")
ax.set_ylabel("value")
ax.set_title("Modulated sine curve and its Hilbert transform")
ax.legend(frameon=False)

# Make all spines/background clean
for spine in ax.spines.values():
    spine.set_alpha(0.8)

plt.tight_layout()

# Show plot
plt.show()

# ------------------------------------------------------------
# Optional: save as transparent PNG
# ------------------------------------------------------------
fig.savefig("modulated_sine_hilbert.png", dpi=300, transparent=True, bbox_inches="tight")
```
